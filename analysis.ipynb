```python
import pandas as pd
df = pd.read_csv('deliveries.csv')
df.head()
```

```python
top_batsmen = df.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(5)
print("--- Top 5 Batsmen ---")
print(top_batsmen)

import matplotlib.pyplot as plt


top_batsmen.plot(kind='bar', color='gold', edgecolor='black')
plt.title('Top 5 Batsmen by Total Runs')
plt.xlabel('Batsman')
plt.ylabel('Total Runs')
plt.xticks(rotation=45)
plt.show()
```
